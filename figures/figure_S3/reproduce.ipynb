# Figure S3
The CSV contains the final survivor population used directly for the plot.


```python
from pathlib import Path
import sys
from IPython.display import display
HERE = Path.cwd()
if not (HERE / "data").is_dir():
    raise FileNotFoundError("Open the notebook from its figure directory.")
sys.path.insert(0, str(HERE.parent / "shared"))
DATA = HERE / "data"
OUTPUT = HERE / "output"
```

```python
import json
import pandas as pd
from property_panels import summarize, render
COHORT = 'gt'
population = pd.read_csv(DATA / 'gt_final_unique.csv', keep_default_na=False)
groups, summary = summarize(population, COHORT)
layout = json.loads((HERE.parent / "shared/property_layout.json").read_text())
width = layout["textwidth_tex_points"] / layout["tex_points_per_inch"] * layout["width_multipliers"][COHORT]
fig, axes = render(groups, COHORT, width)
display(fig)
```

```python
OUTPUT.mkdir(exist_ok=True)
for extension in ("png", "pdf"):
    fig.savefig(OUTPUT / ("distribution_test." + extension), dpi=600, facecolor="white")
pd.DataFrame(summary).to_csv(OUTPUT / "property_summary.csv", index=False)
print(OUTPUT)
```
